## Business Insight

### 1.Ingredient Analysis

```python
ingrediant=(df['pizza_ingredients'].str.
    split(',').
    explode().
    str.strip().
    value_counts().
    reset_index().
    rename(columns={'index':'count','pizza_ingredients':'ingrediants'})
           )
print(ingrediant.head())
```

### 2.Daily Trend--Total Order

```python
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True)
df['day_name']=df['order_date'].dt.day_name()
week_day=['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
df['day_name']=pd.Categorical(df['day_name'],categories=week_day,ordered=True)
order=df.groupby('day_name',observed=False)['order_id'].nunique()
ax=order.plot(kind='bar',figsize=(8,5),color='green',edgecolor='black')
plt.title("Total Orders by Day of Week")
plt.xlabel('Days')
plt.ylabel('Orders')
plt.xticks(rotation=45)

for i, val in enumerate(order):
    plt.text(i, val+20,str(val),ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 3.Daily Trend-Total Revenue

```python
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True)
df['day_name']=df['order_date'].dt.day_name()
week_day=['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
df['day_name']=pd.Categorical(df['day_name'],categories=week_day,ordered=True)
order=df.groupby('day_name',observed=False)['total_price'].sum()
ax=order.plot(kind='bar',figsize=(8,5),color='blue',edgecolor='black')
plt.title("Total Revenue in a Day")
plt.xlabel('Day')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

for i,val in enumerate(order):
    plt.text(i,val+20,f'${val:,.2f}',ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 4.Hourly Trend--Total Order

```python
df['order_time']=pd.to_datetime(df['order_time'],format='%H:%M:%S')
df['order_hour']=df['order_time'].dt.hour
order=df.groupby('order_hour',observed=False)['order_id'].nunique()
ax=order.plot(kind='bar',figsize=(15,5),color='#008080',edgecolor='black')
plt.title("Total Order BY Hours")
plt.xlabel("Hour")
plt.ylabel("Orders")
plt.xticks(rotation=0)

for i,val in enumerate(order):
    plt.text(i,val+20,str(val),ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5.Hourly Trend--Revenue

```python
df['order_time']=pd.to_datetime(df['order_time'],format='%H:%M:%S')
df['order_hour']=df['order_time'].dt.hour
order=df.groupby('order_hour',observed=False)['total_price'].sum()
ax=order.plot(kind='bar',figsize=(15,5),color='#4169E1',edgecolor='black')
plt.title("Revenue By Hours")
plt.xlabel("Hour")
plt.ylabel("Revenue")
plt.xticks(rotation=0)

for i,val in enumerate(order):
    plt.text(i,val+20,f'${val :,.2f}',ha='center',va='bottom',fontsize=9,fontweight='bold')

plt.tight_layout()
plt.show()
```

### 6. Monthly Trend--Total Orders

```python
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True)
df['month']=df['order_date'].dt.month_name()
months_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]
df['month']=pd.Categorical(df['month'],categories=months_order,ordered=True)
orders=df.groupby('month',observed=False)['order_id'].nunique()
plt.figure(figsize=(15,5))
plt.fill_between(orders.index,orders.values,color='orange',alpha=0.6)
plt.plot(orders.index,orders.values,color='black',linewidth=2,marker='o')
plt.title('Monthly Trends by Total Orders')
plt.xlabel('Month')
plt.ylabel('No Of Orders')
plt.xticks(rotation=45)

for i,val in enumerate(orders):
    plt.text(i,val+20,str(val),ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 7.Monthly Trend--Total Revenue

```python
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True)
df['month']=df['order_date'].dt.month_name()
months_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]
df['month']=pd.Categorical(df['month'],categories=months_order,ordered=True)
order=df.groupby('month',observed=False)['total_price'].sum()
plt.figure(figsize=(15,7))
plt.fill_between(order.index,order.values,color='blue',alpha=0.6)
plt.plot(order.index,order.values,color='black',linewidth=2,marker='o')
plt.title('Monthly Trend For Total Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
for i,val in enumerate(order):
    plt.text(i,val+1500,f'${val:,.2f}',ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 8.% Sales by Pizza Category

```python
sale=df.groupby('pizza_category')['total_price'].sum()
sale_pct=sale/sale.sum()*100
plt.figure(figsize=(5,5))
colors=plt.get_cmap('tab20').colors
plt.pie(sale_pct,labels=sale_pct.index,autopct='%1.2f%%',startangle=90,colors=colors,wedgeprops={'edgecolor':'black','width':0.4})
plt.title('Percentage Sales by Pizza Category')
plt.show()
```

### 9. % Sales by Pizza Size and Pizza Category

```python
pivot=df.pivot_table(index='pizza_category',columns='pizza_size',values='total_price',aggfunc='sum',fill_value=0)
sale_pct=pivot/pivot.sum().sum()*100
plt.figure(figsize=(10,5))
sns.heatmap(sale_pct,annot=True,fmt='.1f',cmap='YlOrRd',linewidth=0.5)
plt.title("Percentage Sales by Pizza Category and Pizza Size")
plt.xlabel('Pizza Category')
plt.ylabel('Pizza Size')
plt.show()
          
```

### 10. Total Pizza Sold by Category

```python
sale=df.groupby('pizza_category')['quantity'].sum()
ax=sale.plot(kind='bar',figsize=(5,4),color='blue',edgecolor='black',alpha=0.6)
plt.title("Total Sold Pizza by Category")
plt.xlabel("Pizza Category")
plt.ylabel("Sold")
plt.xticks(rotation=0)
for i,val in enumerate(sale):
    plt.text(i,val+20,str(val),ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```

### 11.Top 5 Pizza Sold

```python
sale=df.groupby('pizza_name')['quantity'].sum()
top=sale.sort_values(ascending=False).head(5)
ax=top.plot(kind='bar',figsize=(10,6),color='grey',edgecolor='black',alpha=0.6)
plt.title("Top 5 Pizza Ordered")
plt.xlabel("Pizza")
plt.ylabel("Sold")
plt.xticks(rotation=45)
for i,val in enumerate(top):
    plt.text(i,val+20,str(val),ha='center',va='bottom',fontsize=9,fontweight='bold')
plt.tight_layout()
plt.show()
```
